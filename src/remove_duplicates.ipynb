```python
def getrowfromimage(fp):
    # img = cv2.imread(fp)
    # source_img = cv2.cvtColor(img,cvs.IMREAD_GRAYSCALE)
    # source_vec = source_img[0,:]

    with Image.open(fp) as img:
        source_img = np.array(img)
        print(source_img.shape)
        _vec = source_img[0,:]
        min = np.min(_vec)
        max = np.max(_vec)
        source_vec = _vec*(-1)+max

    return source_vec
        
```
